```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences=['the glass of milk',
'the glass of juice',
'the cup of tea',
'I am a good boy',
'I am a good developer',
'understanding the meaning of words is important','your videos are good',]
```

```python
#   Defining the vocabulary size
voc_size=10000
```

```python
#One Hot Representation
one_hot_repr=[one_hot(word, voc_size) for word in sentences]
one_hot_repr
```

```python
#Word embedding representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sentence_length=8
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sentence_length)
print(embedded_docs)
```

```python
#Feature representation
dimensions=10
model=Sequential()
model.add(Embedding(voc_size,dimensions,input_length=sentence_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
